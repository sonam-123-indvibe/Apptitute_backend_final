```python
import pandas as pd

df = {
    "name" : ["Kunal" ,"Dev"],
    "age" : [22,21]
}

df2 = {
    "name" : ["Kunal" ,"Dev"],
    "age" : [22,21]
}

dataf = pd.DataFrame(df)
dataf2 = pd.DataFrame(df2)

print(dataf)
print(dataf2)

finaldf =  pd.concat([dataf , dataf2] , ignore_index = True)

print(finaldf)
```

```python
import pandas as pd

df = {
    "name" : ["Kunal" ,"Dev"],
    "id" : [1,2]
}

df2 = {
    "city" : ["Indore" ,"Dewas"],
    "age" : [22,21],
    "id" : [1,2]
}

dataf = pd.DataFrame(df)
dataf2 = pd.DataFrame(df2)



finaldf =  pd.merge(dataf , dataf2 , on="id" , how="inner")
finaldf2 =  pd.merge(dataf , dataf2 , on="id" , how="left")
finaldf2 =  pd.merge(dataf , dataf2 , on="id" , how="right")

print(finaldf)
```
